```python
import pandas as pd

df = pd.read_csv("dataset.csv")

df["errorBalanceOrig"] = (
    df["oldbalanceOrg"]
    - df["amount"]
    - df["newbalanceOrig"]
)

df["errorBalanceDest"] = (
    df["oldbalanceDest"]
    + df["amount"]
    - df["newbalanceDest"]
)

print(df[[
    "amount",
    "errorBalanceOrig",
    "errorBalanceDest"
]].head())

print("Feature engineering completed!")
```
